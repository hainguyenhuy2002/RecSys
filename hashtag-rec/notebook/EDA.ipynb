```python
with open("data/trava_elite_2W_v2.tweets.json",'r') as rfile:
    data = json.load(rfile)
```

```python
def get_hashtag(message):
    mess = re.split(r'\s+|\n|\[|\]', message)
    hashtags =[]
    for me in mess:
        if "#" in me:
            hashtags.append(me[1:])
    return hashtags
```

```python
hashtags = []
for da in data: 
    hashtag_info = dict()
    hashtag_info['content'] = da['text']
    hashtag_info['author'] = da['author']
    hashtag_info['authorName'] = da['authorName']
    hashtag_info['hashtags'] =get_hashtag(da['text'])
    hashtags.append(hashtag_info)
```

```python
df_hashtags = pd.DataFrame(hashtags)
df_hashtags.iloc[200].content
```

```python
df_dct = df.to_dict()
dct ={}
for i in range(len(df)):
    inner_dct={}
    iso_code = df_dct['iso_code'][i]
    inner_dct['iso_code']=df_dct['iso_code'][i]
    inner_dct['precision']=df_dct['precision'][i]
    inner_dct['symbol']=df_dct['symbol'][i]
    inner_dct['pattern']=df_dct['pattern'][i]
    inner_dct['negative_pattern']=df_dct['negative_pattern'][i]
    dct[f'{iso_code}'] = inner_dct
```

```python
with open("data/currency.json","w") as wfile:
    json.dump(dct, wfile, indent=4)
```
