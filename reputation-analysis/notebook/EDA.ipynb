```python
def get_labels(row):
    if row   <= 20150:
        return 'poor'
    elif 20150<row<= 20500:
        return 'fair'
    elif 20500<row<=21500:
        return 'good'
    elif 21500<row<=25000:
        return 'very good'
    elif row> 25000:
        return 'exceptional'
    
```

```python

aave_profile['score_label'] =aave_profile['score'].apply(lambda row: get_labels((row)))
```

```python
label_df = pd.DataFrame(aave_profile['score_label'].value_counts()).reset_index()
```

## Tags

```python
def get_tags_label(x):
    set_x = set(x)
    user_set = {'aave_users', 'aave_v3_borrowers', 'aave_v2_borrowers', 'aave_v2_depositors', 'aave_v3_borrowers', 'aave_v3_depositors'}
    intersection_set = set_x & user_set
    if len(intersection_set)>=1:
        return 'aave_user'
    
    else:
        return 'aave_only_holder'
```

```python
aave_profile_cleaned = aave_profile.dropna(subset=['tags','elite'])
aave_profile_cleaned['tags'] = aave_profile_cleaned['tags'].apply(eval)
#aave_profile_tags  = aave_profile_cleaned.explode("tags")
aave_profile_cleaned['tags_label'] = aave_profile_cleaned['tags'].apply(lambda x: get_tags_label(x))
```

Price of aave token in 1_3: 110.53 USD \
Price of stkaaave token in 1_3: 108.17 USD

```python
def extract_token_balance_USD(row, tokens):
    value =0
    for token in tokens:
        if token in row:
            if token =='0x7fc66500c84a76ad7e9c93437bfc5ac33e2ddae9':
                value = row[token] *110.53
            elif token =='0x4da27a545c0c5b758a6ba100e3a049001de870f5':
                value = row[token] *108.17
    if value !=0:
        return value
    else: 
        return None

def get_inactive_user(row):
    if row['Aave_balance'] <= 100:
        return 'inactive'
    else:
        return row['score_label']
```

```python
aave_holders = aave_profile_cleaned[aave_profile_cleaned['tags_label']=='aave_only_holder']
aave_holders_cleaned = aave_holders.dropna(subset=['tokens'])

aave_holders_cleaned['Aave_balance'] = aave_holders_cleaned['tokens'].apply(lambda row: extract_token_balance_USD(eval(row), ['0x7fc66500c84a76ad7e9c93437bfc5ac33e2ddae9','0x4da27a545c0c5b758a6ba100e3a049001de870f5']))
aave_holders_cleaned = aave_holders_cleaned.dropna(subset=['Aave_balance'])
aave_holders_cleaned['score_labels'] = aave_holders_cleaned.apply(get_inactive_user, axis=1)

#holder_score = pd.DataFrame(aave_holders.score_label.value_counts()).reset_index()
```
